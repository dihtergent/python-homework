```python
class Dog():
    def __init__(self,name,gender,gene,age,weight,color):
        self.name = name
        self.gender = gender
        self.gene = gene
        self.age = age
        self.weight = weight
        self.color = color
    def sit(self):
        print("{} sit".format(self.name))
    def sleep(self):
        print("{} sleep weight lost ".format(self.name))
    def eat(self):
        print("{} sit".format(self.name))
    def showinfo(self):
        print("""
        ชื่อ: {}
        เพศ: {}
        พันธุ์: {}
        อายุ: {}
        นํ้าหนัก: {}
        สี: {}
        """.format(self.name,self.gender,self.gene,self.age,self.weight,self.color))
```

```python
wed = Dog ("josh","male","BlueArchive",36,40,"PinkPurple")
wed.showinfo()
```

```python
class student():
    def __init__(self,SID,CID,Fname,Lname,AGE,GENDER):
        self.sid = SID
        self.id = None
        self.id = self.set_id(CID)
        self.Fname = Fname
        self.Lname = Lname
        self.age = AGE
        self.gender = GENDER
    def set_id(self,CID):
        if not((len(CID)==13)and(self.id == None)):
            print("error")
            return
        else : return CID

    def set_age(self,AGE):
        if (AGE > self.age):
            self.age = AGE
    def show_info(self):
        print("""
        รหัสนักศึกษา: {}
        รหัสบัตรประชาน: {}
            ชื่อ: {}
        นามสกุล: {}
            อายุ: {}
            เพศ: {}
        """.format(self.sid,self.id,self.Fname,self.Lname,self.age,self.gender))
```

```python
nu = student("UC 95","1339900857963","Hi","Nu",3,"N")
nu.show_info()
nu.set_age(5)
nu.show_info()
nu.set_age(4)
nu.show_info()
```
